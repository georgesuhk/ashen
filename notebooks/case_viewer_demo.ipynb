# Case viewer: demo on made-up files

Tries every view of `ashen.viewer` without JOREK or the HPC. The first cell writes a
small made-up campaign into `demo_campaign/` next to this notebook: a `site.toml`, a
template, and one run folder whose inputs are prepared by the real `run_jorek` code
from a `ffprime_method = "q_li"` shotfile.

**The run's "JOREK output" is invented**, not physics: an equilibrium that is the
cylinder model with q0, l_i and q_edge a little off the request (+0.05, +0.06, -0.12),
and modes that grow and then saturate. It only has the shapes and file formats the
viewer reads.

Needs `ipywidgets`, `matplotlib` and `h5py`: `pip install ipywidgets ipykernel h5py`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))   # ashen's src/, relative to this notebook

from ashen import demo, viewer

RUN = demo.make_demo_campaign("demo_campaign")
print(RUN)
print(sorted(p.name for p in RUN.iterdir()))
```

## Current profile from q0, l_i, q_edge

Things to try:
1. Move the sliders. The red dashed line is the pretend JOREK equilibrium, which was
   made from the shotfile's values when the first cell ran, so it stays put.
2. Push `q0` far down: outside the reachable window the plot is replaced by the window.
3. **Save to shotfile**, then look at `demo_campaign/demo_shot/q_li_demo/shotfile.py`.
4. **Regenerate inputs**, then look at `j_prof.dat` and `ffprime_prof.dat` in the same
   folder: their header and values follow the saved numbers.
5. Re-run the first cell: the pretend equilibrium is rebuilt from the saved shotfile,
   again a little off, as a real `--run_eq` would be.

```python
viewer.profile_tuner(RUN, step=0)
```
